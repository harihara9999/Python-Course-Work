```python
# in while loop u have to increment r decrement the value if not it ewill be infinite loop
# initialization, condition, increment/decrement
```

```python
# example - 1

i = 1
while i <= 10:
    print(i)
    i +=1
```

```python
# Example number 2 : Break Statment
# break statement helps to come out of the loop

i = 1
while i <= 10:
    print(i)
    if i == 5:
        break   # break will help u to comeout of loop when i ==5
    i += 1
```

```python
i = 1
while i <= 10:
    i += 1
    print(i)
    if i == 5:
        break   # break will help u to comeout of loop when i ==5
```

```python
# Example number 3 : Continue Statment (if continue satement is true it skips that values )
# Continue will take the control to the start of the loop
# if continue will execute then further statement next lline will not execute
# continue will take u to starting of the looop

i = 1
while i < 10:
    i += 1
    if i < 5:
        continue
    print(i)
```

```python
# example : else statement with while loop
# else statement executes only if loop is successfully completed
# we use else to check if the loop is successfully completed r not

i = 1
while i <=10:
    print(i)
    i += 1

else:     # to check if the loop is successfully completed r not
    print("loop is successfully completed")
```

# for loop

```python
animals = ["pulli", "pilli", "kukka", "pandhi", "kothi"]
for i in animals:
    print(i)
```

```python
# for string type it give chachater by chachater
food = "chicken_bri"
for i in food:
    print(i)
```

```python
# for with the range()
# range() fun is used to generated list of the elements
#syntax of range: range(start, end, step)

for i in range(10):   #start=0, end = end -1, step =1
    print(i)
```

```python
for i in range(4,10):  #start=4, end = 10 -1, step =1
    print(i)
```

```python
# for loop automatically increment the i value

for i in range(4,10,2):      #start=4, end = 10 -1, step =2
    print(i)
```

```python
# example : break statement - to cmg out of the loop

for num in range(1,6):
    if num == 3:         # as n==3 so it will break at 3
        break
    print(num)
```

```python
#  example : continue statement - take the control to the start on the loop

for num in range(1,6):
    if num == 3:         # as n==3 so it will take control to start of loop at 3
        continue
    print(num)
```

```python
# example : nested loop
# for every value of outer loop, inner loop will completely execute

for i in range(1,4):
    for j in range(1,3):  # for 1 value of outer loop complete inner loop will execute
        print(i,j)
```

```python
# example : else statement - executes if loop complete execution successfully

for i in range(5):
    if i == 2:
        continue
    elif i == 4:
        break
    else:
        print(i)
    pass
else:
    print("loop completed successfully")
```
