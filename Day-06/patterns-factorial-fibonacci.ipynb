# Pattern

```python
for i in range(5):
    print("*", end= " ")
```

```python
for i in range(5):
    for j in range(5):
        print("*", end= " ")
    print()
```

```python
for i in range(1,6):
    for j in range(i):
        print("*", end = " ")
    print()
```

```python
for i in range(1,6):
    for j in range(6-i):
        print("*", end = " ")
    print()
```

```python
# other way
for i in range(5,0,-1):
    for j in range(i):
        print("*", end = " ")
    print()
```

```python
n= 5
for i in range(n,0,-1):
    print(" " * i + "* " * (n-i+1))
```

```python
n= 5
for i in range(n-1,0,-1):
    print(" " * i + "* " * (n-i+1))
```

```python
# Example: Factorial

n = int(input())
fact = 1
for i in range(1, n+1):
    fact = fact * i
print(fact)
```

```python
# Example: Fibonacci series
n = int(input())

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a+b
```
