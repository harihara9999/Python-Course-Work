# Input in Python

```python
#input function is used to take input from the user


val = input("Enter your phone number")

print(val)
print(type(val)) #By default input function recieve/return the data in string format
```

```python
val = int(input("Enter your phone number"))

print(val, type(val)) # string format can be coverted into other type by using typecasting method
```

```python
# recieving multiple inputs we use split function( .split() )

x1, x2, x3 = input("Enter 3 numbers").split() # by default give multiple inputs by giving space
print(x1, x2, x3)
```

```python
x1, x2, x3 = input("Enter 3 names").split(",") #if u want to use special char mention that in bracket()
print(x1)
```

```python
#map()

x1, x2, x3 = map(int, input("Enter 3 num").split()) #for typecasting we use map
print(x1, x2, x3)
```

```python
x1, x2, x3 = map(float, input("Enter 3 names").split()) #for typecasting we use map
print(x1, x2, x3)
```

```python
x = list(map(int, input("Enter 3 names").split())) #if we can to create list
print(x)
```

# output in python

```python
# 1. concatenation using commas
# 2. string concatenayion using +
# 3. str.format() method
# 4. f.string method (python 3.6+ version)
# 5. Escape sequences
# 6. End and sep parameters in print()
```

```python
# 1. concatenation using commas

Hero_Name = 'prabhas'
age = 40
fee = 100
print(Hero_Name, age, fee)
```

```python
# 2. string concatenation using +

Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100

print(Hero_Name + " " + Heroine_Name + str(fee)) # + concate can happen only using str if it is not string convert it into str()
```

```python
# 3. str.format() method #we don't use now 
Hero_Name = 'NTR'
age = 40

print("Name: {}  age: {}".format(Hero_Name,age)) # {} will take variable we can also mention by giving number {0} {1}
```

```python
# 4. f.string method (python 3.6+ version) print(f"")
Hero_Name = 'NTR'
age = 40

print(f"Name: {Hero_Name} , Age:{age}") # if u want to print value of variable should me written in the {}
```

```python
# 5. Escape sequences
Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100

print(f"Name: {Hero_Name},\nHeroine_Name:{Heroine_Name},\tFee:{fee}\n") #if u want to get in next line use \n and we u want tab space use \t
print(f"Name: \"{Hero_Name}\",\nHeroine_Name:{Heroine_Name},\tFee:{fee}\n") # \" used for escape sequence used to include double quote
```

```python
# 6. End and sep parameters in print() # by default end = "\n" & sep = " "(space)
Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100

print(Hero_Name, end = "%") # by default it go to next line end is making sure that wht u want to do after printing
```

```python
Hero_Name = 'Prabhas'
Heroine_Name = 'Anushka'
fee = 100
print(Hero_Name,Heroine_Name,fee, sep = "@")
```

```python
# end parameter is used  to decide how the print statement will END
# sep parameter decides how the multiple values will be separated
```
