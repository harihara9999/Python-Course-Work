# Updating list items

```python
list_politician = ["vijay", "CBN", "Pavan", "JaiLalita", "Balayya", "Anushka"]

#update single element
list_politician[1] = "Dance With"

print(list_politician)
```

```python
# update multiple elements: update more elements with less
list_politician[3:5] = ["Dance With"]
print(list_politician)
```

```python
#update less elements with more
list_politician[1:4] = ["Rahul", "Kejriwal", "Yogi", "Mamta"]
print(list_politician)
```

```python
print(list_politician  + ["Akshara"])
```

```python
list_politician[-2] = "Dance with"
print(list_politician)
```

```python
list_politician[-5:-1] = ["Dance with"]
print(list_politician)
```

# insert items in list

#### append(), extend(), insert()

```python
l2 = []

#append() function is used to insert the single 
l2.append(10)
print(l2)
```

```python
l2.append(20)
print(l2)
```

```python
# extend() : this function is used to insert multiple elements in the last

l2.extend([10, 50, 60, 80, 30])
print(l2)
```

```python
# insert() can be used to insert at specific index
l2.insert(4, "Dinesh")
print(l2)
```

# Delete item from list (remove(),pop(),clear(),del)

```python
food_list = ["Papu", "Biryani", "Fired rice", "Sambar","Payasum","Curd rice"]

#remove(): delete the item with given name
food_list.remove("Fired rice") # when u want to remove items by name of items
print(food_list)
```

```python
food_list.pop(4) # we use pop with and without index , when we use without index it delete the last index
print(food_list)
```

```python
# pop() without index it delete the last index
food_list.pop()
print(food_list)
```

```python
food_list = ["Papu", "Biryani", "Fired rice", "Sambar","Payasum","Curd rice"]

# delete muultiple elements together
del food_list[1:4]
print(food_list)
```

```python
# clear() is used to delete all the items from the list
food_list.clear()
print(food_list)
```

```python
# del is used in two ways to delete multiple elements together in range and to delete whole list
del food_list    # del is keyword it not a function
```

# popular list inbuilt functions

```python
Age_list = [22,23,22,20,19,18,21,16,19,20,99]

print(len(Age_list))   # to find out the length of list
print(min(Age_list))    # to find max value of the list
print(max(Age_list))    # to find min value of the list
print(sum(Age_list))    # to find sum of values

Age_list.reverse()       # reversing the list
print(Age_list)

# In python sort() function use the TIM sort -> (Merge + Insert) -> time complexity: O(nlogn)

Age_list.sort()       # sorting in asending order
print(Age_list)

Age_list.sort(reverse = True)       # sorting in desending order
print(Age_list)
```

```python
# list() :- we can create empty list and we can convert other to list

# covert the tuple to list
t1 = (10,20,30,40,50)
s1 = {80,90,55,120,12,50}
l2 = list(t1)
l3 = list(s1)
print(l2)
print(l3)

value = range(1,10)  # we use range() to get the certain range of items
l4= list(value)
print(l4)

l5 = list(input("Enter names").split())  # we use list() to get input from user and convert it to list
print(l5)
```

# list comprehenion

```python
# creating the list in a single line using a loop

l1 = [i for i in range(1,11)]
print(l1)
```

```python
l2 = [i**2 for i in range(1,11)]
print(l2)
```

```python
l3 = [i for i in range(1,11) if i % 2 == 0] 
print(l3)

# for i in range(1,11):
#    if i%2==0:
#        print(i)    # this one is written in single line
```

```python
l4 = [i for i in range(1,11) if i % 2 == 0 if i%3==0] 
print(l4)
```

```python
l5 = [i**2 if i%2== 0 else i+2 for i in range(1,11)]
print(l5)
```

```python
l5 = [i**2 if i%2== 0 else i+2 for i in range(1,11) if i > 5 and i < 9]  # here we can't use else of right hand but u can use OR,AND
print(l5)
```

```python
l6 = [[i for i in range(1,5)] for j in range(1,3)]     # 2d list
print(l6)
```

```python
l6 = [[[i for i in range(1,5)] for j in range(1,3)]for k in range(1,3)]    # two 2d list inside a 3d list
print(l6)
```
