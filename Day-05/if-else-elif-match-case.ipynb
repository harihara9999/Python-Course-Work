# conditional statements in python

```python
# Example 1     Structure of if statements : if(statement):
a = 10
if (a % 2 == 0):
    print(a)
```

```python
if (a % 2 == 0): print(a) # we can write in single statement also
```

```python
# Example 2     Structure of if-else statements
a = 10
if a % 2 == 0:
    print("Even")
else:
    print("ODD")
```

```python
# Ternary condition (writing if-else condition in single statement)

print("Even") if a % 2 == 0 else print("ODD")
```

```python
a=15
print("Hello") if a%3 == 0 else print("World") # first write output required for if then if condition then else with its output
```

```python
# Example 3     Structure of elif statements(if one condition fails then we go to next condition)

Biryani = input("Famous Hyderabadi biryani")

if Biryani == "Paradise":
    print("Spicy & tasty")
elif Biryani == "Shahdab":
    print("double masala")
elif Biryani == "road side":
    print("budget friendly")
else:
    print("PG Biryani")
```

```python
# replace elif with else: if
Biryani = input("Famous Hyderabadi biryani")

if Biryani == "Paradise":
    print("Spicy & tasty")
else:
    if Biryani == "Shahdab":
        print("double masala")
    else:
        if Biryani == "road side":
            print("budget friendly")
        else:
            print("PG Biryani")
```

```python
Biryani = input("Famous Hyderabadi biryani")
print("Spicy & tasty") if Biryani == "Paradise" else print("double masala") if Biryani == "Shahdab" else print("budget friendly") if Biryani == "road side" else print("PG Biryani")
```

# match case

```python
# Match case is used to replace multiple if else condition
# Match case is more powerful than switch case because match case works with variables & data Structurs
```

```python
# Example 1
num = int(input())

match num:
    case 10:
        print("you entered 10")
    case 20:
        print("you entered 20")
    case 30:
        print ("you entered 30")
    case _:
        print("something else")
```

```python
# Example 2 : passind data structure to match case

data = [1,2,3,4]
match data:
    case [4,3,2,1]:
        print("Hello")
    case [1,2,3,4]:
        print("world")
```

```python
# example
Biryani = input("Famous Hyderabadi biryani")
match Biryani:
    case "Paradise":
        print("Spicy & tasty")
    case "Shahdab":
        print("double masala")
    case "road side":
        print("budget friendly")
    case _:
        print("something else")
```

```python
#Example 3: if condition with match case

val = int(input())

match val:
    case x if x > 0:
        print("it is positive number")
    case y if y < 0:
        print("it is negative number")
    case _:
        print("it is a zero")
```
